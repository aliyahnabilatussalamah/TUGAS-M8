```python
# Soal No. 6
pelamar = [("Budi", 3.2, 5), ("Siti", 3.6, 3), ("Dodi", 3.8, 1)]
for nama, ipk, smt in pelamar:
    if ipk >= 3.5 and smt > 2:
      print("Nama Pelamar:",nama,"| Status: Diterima")
    else:
        print("Nama Pelamar:",nama,"| Status: Ditolak")
```

```python
# Soal NO. 7
karyawan = [("Joko", "IT", 7000), ("Rina", "HR", 5000), ("Toni", "Data", 8000), ("Lia", "IT", 6500)]

total_gaji= 0
for nama, depart, gaji in karyawan:
    if depart == "IT" or "Data":
        total_gaji += gaji

print(f"Total Gaji Karyawan Departemen IT dan Data : Rp{total_gaji:,}")
```
